```python
def set_seed(seed):
    random.seed(seed)
    np.random.seed(seed)
    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

set_seed(199)
```

```python
import os
from PIL import Image
from torch.utils.data import Dataset
from torchvision import transforms

class CubeCylinderDataset(Dataset):
    def __init__(self, root_dir, transform=None):
    
        self.root_dir = root_dir
        self.transform = transform
        self.samples = []
        
        cube_path = os.path.join(root_dir, "fCube_Images")
        cylinder_path = os.path.join(root_dir, "fCylinder_Images")
        

        for file in os.listdir(cube_path):
            if file.lower().endswith((".png", ".jpg", ".jpeg", ".bmp")):
                self.samples.append(
                    (os.path.join(cube_path, file), 0)
                )
        
        for file in os.listdir(cylinder_path):
            if file.lower().endswith((".png", ".jpg", ".jpeg", ".bmp")):
                self.samples.append(
                    (os.path.join(cylinder_path, file), 1)
                )
    
    def __len__(self):
        return len(self.samples)
    
    def __getitem__(self, idx):
        img_path, label = self.samples[idx]
        image = Image.open(img_path).convert("RGB")
        
        if self.transform:
            image = self.transform(image)
        
        return image, label
    
    
```

```python
import os
from PIL import Image

# ---------------------------------------------------------
# Configurações da Animação
# ---------------------------------------------------------

# Nome da pasta que contém as imagens
FOLDER_NAME = 'gif_change'

# Nome base das imagens (ex: 'p' para 'p1.png', 'p2.png'...)
IMAGE_BASE_NAME = 'p'

# Extensão das imagens (com o ponto)
IMAGE_EXTENSION = '.png'

# Quantidade total de imagens
NUM_IMAGES = 10

# Duração de cada quadro (frame) em milissegundos
# 200 ms é uma boa velocidade inicial. Aumente para mais devagar, diminua para mais rápido.
DURATION_PER_FRAME = 200 

# Nome do arquivo GIF de saída
OUTPUT_GIF_NAME = 'animacao_pontos.gif'


# ---------------------------------------------------------
# Script Principal
# ---------------------------------------------------------

def create_animated_gif():
    # Lista para armazenar os objetos de imagem do Pillow
    frames = []

    # 1. Verificar se a pasta existe
    if not os.path.exists(FOLDER_NAME):
        print(f"❌ Erro: A pasta '{FOLDER_NAME}' não foi encontrada.")
        print(f"Certifique-se de que ela existe no mesmo diretório deste script e contém as imagens p1.png até p10.png.")
        return

    print(f"📂 Lendo imagens da pasta '{FOLDER_NAME}'...")
    
    # 2. Ler as imagens em ordem (p1 até p10)
    for i in range(1, NUM_IMAGES + 1):
        # Constrói o caminho completo para cada imagem: 'gif_change/p1.png', etc.
        image_path = os.path.join(FOLDER_NAME, f"{IMAGE_BASE_NAME}{i}{IMAGE_EXTENSION}")
        
        # Verificar se o arquivo de imagem existe antes de tentar abrir
        if os.path.exists(image_path):
            try:
                # Abre a imagem e converte para RGB para garantir compatibilidade com o formato GIF
                # (isso é importante se você tiver imagens RGBA com transparência)
                img = Image.open(image_path).convert("RGB")
                
                # Adiciona a imagem aberta à nossa lista de quadros
                frames.append(img)
                print(f"  ✅ Imagem '{image_path}' carregada.")
            except Exception as e:
                print(f"  ❌ Erro ao abrir a imagem '{image_path}': {e}")
        else:
            print(f"  ⚠️ Erro: A imagem '{image_path}' não foi encontrada.")

    # 3. Verificar se alguma imagem foi carregada com sucesso
    if not frames:
        print("❌ Nenhuma imagem válida foi carregada. O GIF não pôde ser criado.")
        return

    # 4. Criar e salvar o GIF animado
    print(f"\n🎬 Criando o GIF animado '{OUTPUT_GIF_NAME}'...")
    
    try:
        # Pega a PRIMEIRA imagem como base para a animação
        # e usa o método .save() nela.
        frames[0].save(
            OUTPUT_GIF_NAME,               # Nome do arquivo de saída
            save_all=True,                  # Salvar todas as imagens, não apenas a primeira
            append_images=frames[1:],       # Lista das imagens restantes para anexar
            duration=DURATION_PER_FRAME,    # Tempo de cada quadro em milissegundos
            loop=0                          # Número de loops (0 significa loop infinito)
        )
        print(f"✅ GIF '{OUTPUT_GIF_NAME}' criado com sucesso no mesmo diretório!")
    except Exception as e:
        print(f"❌ Erro ao salvar o GIF: {e}")

if __name__ == "__main__":
    # Chama a função principal
    create_animated_gif()
    print("\nSaindo do script.")
```
